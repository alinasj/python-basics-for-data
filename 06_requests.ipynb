## GET a public API

```python
# GET a public API
import requests

url = "http://api.open-notify.org/astros.json"
response = requests.get(url, timeout=10)

print("status_code -->", response.status_code)  # 200 means OK
print("type(content) -->", type(response.content))  # <class 'bytes'>
print("type(json)    -->", type(response.json()))   # <class 'dict'>
```

## Parse JSON

```python
# .json() turns the body into a dict you can index
data = response.json()
print("keys -->", data.keys())
print("people in space -->", data["number"])
print("first person -->", data["people"][0])
```

## Walk nested JSON

```python
# Walk nested JSON with a function
def print_astronauts(payload: dict) -> None:
    print("people in space:", payload["number"])
    for person in payload["people"]:
        print(" ", person["name"], "on", person["craft"])


print_astronauts(response.json())
```

## Handle errors

```python
# Always check for failure before you parse
bad = requests.get("https://httpbin.org/status/404", timeout=10)
print("bad status -->", bad.status_code)  # 404

try:
    bad.raise_for_status()
except requests.HTTPError as err:
    print("raise_for_status caught -->", err)
```

## Save JSON

```python
# Save JSON to a file — this is how you hand data to pandas later
import json
from pathlib import Path

out = Path("astronauts.json")
out.write_text(json.dumps(response.json(), indent=2))
print("wrote", out.resolve())
```
