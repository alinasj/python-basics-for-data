## Loops

<a id="while-loop"></a>

### while loop

```python
# while loop — repeats until the condition is False
x = 0
while x < 3:
    print("x =", x)
    x += 1
# x = 0
# x = 1
# x = 2
```

<a id="for-loop-with-range"></a>

### for loop with range

```python
# for loop with range
for i in range(3):
    print(i)
# 0
# 1
# 2

print("range(3, 6)    -->", list(range(3, 6)))     # [3, 4, 5]
print("range(3, 8, 2) -->", list(range(3, 8, 2)))  # [3, 5, 7]
```

<a id="for-loop-over-a-list-string-tuple-and-dict"></a>

### for loop over a list, string, tuple, and dict

```python
# for loop over a list, string, tuple, and dict
for item in [1, 2, 3]:
    print("list item -->", item)

for char in "Hi":
    print("char -->", char)

for item in (10, 20):
    print("tuple item -->", item)

person = {"name": "John", "age": 20}
for key, value in person.items():
    print(key, "-->", value)
```

<a id="nested-loops"></a>

### Nested loops

```python
# Nested loops
for x in range(2):
    for y in range(2):
        print(f"({x}, {y})")
# (0, 0)
# (0, 1)
# (1, 0)
# (1, 1)
```

<a id="break-stops-the-loop-continue-skips-this-round"></a>

### break stops the loop; continue skips this round

```python
# break stops the loop; continue skips this round
for n in range(5):
    if n == 3:
        break
    print("break demo -->", n)
# 0, 1, 2

for n in range(5):
    if n == 2:
        continue
    print("continue demo -->", n)
# 0, 1, 3, 4
```

## Functions

<a id="a-function-packages-a-task-so-you-can-reuse-it"></a>

### A function packages a task so you can reuse it

```python
# A function packages a task so you can reuse it
def greet(name):
    print(f"Hello, {name}!")


greet("John")  # Hello, John!
```

<a id="default-parameters"></a>

### Default parameters

```python
# Default parameters
def greet(name="World"):
    print(f"Hello, {name}!")


greet()         # Hello, World!
greet("Alina")  # Hello, Alina!
```

<a id="return-gives-a-value-back-to-the-caller"></a>

### return gives a value back to the caller

```python
# return gives a value back to the caller
def increment(number, by=1):
    return number + by


print(increment(2))        # 3
print(increment(2, by=5))  # 7
```

<a id="args-collects-extra-positional-arguments-into-a-tuple"></a>

### *args collects extra positional arguments into a tuple

```python
# *args collects extra positional arguments into a tuple
def multiply(*numbers):
    total = 1
    for n in numbers:
        total *= n
    return total


print(multiply(2, 3, 4))  # 24
```

<a id="kwargs-collects-extra-named-arguments-into-a-dict"></a>

### **kwargs collects extra named arguments into a dict

```python
# **kwargs collects extra named arguments into a dict
def save_user(**user):
    print(user)


save_user(id=1, name="John", age=22)
# {'id': 1, 'name': 'John', 'age': 22}
```

<a id="lambda-a-tiny-one-line-function"></a>

### Lambda: a tiny one-line function

```python
# Lambda: a tiny one-line function
add_one = lambda x: x + 1
print(add_one(1))  # 2

# Same thing as a normal function:
def add_one_named(x):
    return x + 1


print(add_one_named(1))  # 2
```

## Comprehensions

<a id="list-comprehension"></a>

### List comprehension

```python
# List comprehension
print([x + 1 for x in [1, 2, 3, 4, 5]])  # [2, 3, 4, 5, 6]
print([x for x in range(6) if x % 2 == 0])  # [0, 2, 4]
```

<a id="dict-set-and-generator"></a>

### Dict, set, and generator

```python
# Dict, set, and generator
print({x: x + 1 for x in [1, 2, 3]})  # {1: 2, 2: 3, 3: 4}
print({x for x in [1, 1, 2, 2, 3]})   # {1, 2, 3}

gen = (x for x in [1, 2, 3])
print(gen)          # <generator object ...>
print(list(gen))    # [1, 2, 3]  — generators yield values on demand
```
