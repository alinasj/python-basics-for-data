# 10 — Visualization and descriptive stats

```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sales = pd.DataFrame(
    {
        "city": ["Philly", "NYC", "Philly", "NYC", "Philly", "Boston"],
        "amount": [120, 25, 40, 80, 200, 15],
        "n_items": [1, 1, 1, 2, 3, 1],
    }
)
sales
```

## describe

```python
# The mean can mislead when one value is huge
print(sales["amount"].mean())    # pulled up by 200
print(sales["amount"].median())  # more typical
print(sales["amount"].describe())
print(sales["city"].value_counts())
```

## matplotlib

```python
# Bar chart — total amount by city
totals = sales.groupby("city")["amount"].sum()
totals.plot(kind="bar", title="Revenue by city")
plt.ylabel("amount")
plt.tight_layout()
plt.show()
```

## seaborn

```python
# seaborn is for "show me the distribution" and "split by category"
sns.barplot(data=sales, x="city", y="amount", estimator="sum", errorbar=None)
plt.title("Revenue by city")
plt.show()

sns.histplot(sales["amount"], bins=5)
plt.title("Amount distribution")
plt.show()
```

## correlation

```python
# correlation is -1 to 1. It is not causation.
print(sales[["amount", "n_items"]].corr())

sns.scatterplot(data=sales, x="n_items", y="amount", hue="city")
plt.title("Items vs amount")
plt.show()
```
