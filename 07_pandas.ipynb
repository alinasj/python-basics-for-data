## Import pandas

```python
import pandas as pd
```

## Series

```python
# Series — one column with an index
s = pd.Series([1, 3, 5, None, 6, 8], name="n")
print(s)
print("mean -->", s.mean())  # ignores missing values
```

## DataFrame

```python
# DataFrame from a dict of columns
houses = pd.DataFrame(
    {
        "owner": ["Fath", "Srutika", "Aileen", "Fath"],
        "city": ["Philly", "Philly", "NYC", "NYC"],
        "baths": [5, 5, 3, 2],
        "bedrooms": [10, 1, 4, 3],
        "kitchens": [5, 3, 20, 1],
    }
)
houses
```

## Inspect

```python
# Inspect
print(houses.shape)       # (4, 5)  → 4 rows, 5 columns
print(houses.columns.tolist())
print(houses.dtypes)
houses.head()
```

## Filter rows

```python
# Select columns and filter rows
print(houses["owner"])
print(houses[["owner", "baths"]])

philly = houses[houses["city"] == "Philly"]
philly
```

## loc and iloc

```python
# loc (labels) vs iloc (positions)
print(houses.loc[0, "owner"])   # Fath
print(houses.iloc[0, 0])        # Fath
houses.iloc[0:2]
```

## groupby

```python
# groupby — split, apply, combine
houses.groupby("city")[["baths", "bedrooms"]].mean()
```

## Read / write CSV

```python
# Read / write CSV
from pathlib import Path

path = Path("houses.csv")
houses.to_csv(path, index=False)
loaded = pd.read_csv(path)
loaded
```
