```python
import duckdb
import pandas as pd
from pathlib import Path

sales = pd.DataFrame(
    {
        "city": ["Philly", "NYC", "Philly", "NYC", "Boston"],
        "product": ["bike", "lock", "helmet", "lock", "bike"],
        "amount": [120, 25, 40, 80, 90],
    }
)
customers = pd.DataFrame(
    {
        "city": ["Philly", "NYC", "Boston"],
        "region": ["Mid-Atlantic", "Northeast", "Northeast"],
    }
)

Path("sales.csv").write_text(sales.to_csv(index=False))
sales
```

## Load a CSV

```python
# DuckDB can query a CSV without loading it into pandas first
duckdb.sql("SELECT * FROM 'sales.csv'").show()
```

## SELECT and WHERE

```python
duckdb.sql(
    '''
    SELECT city, amount
    FROM sales
    WHERE amount >= 80
    ORDER BY amount DESC
    '''
).show()
# You can query a pandas DataFrame by its variable name.
```

## GROUP BY

```python
duckdb.sql(
    '''
    SELECT city, SUM(amount) AS revenue, COUNT(*) AS orders
    FROM sales
    GROUP BY city
    HAVING SUM(amount) >= 100
    ORDER BY revenue DESC
    '''
).show()
```

## JOIN

```python
duckdb.sql(
    '''
    SELECT c.region, SUM(s.amount) AS revenue
    FROM sales AS s
    LEFT JOIN customers AS c
      ON s.city = c.city
    GROUP BY c.region
    ORDER BY revenue DESC
    '''
).df()  # .df() sends the result back to pandas
```
