## Why classes exist

```python
# Without a class you repeat the same variables for every house
fath_bath, fath_bedroom, fath_kitchen = 5, 10, 5
sru_bath, sru_bedroom, sru_kitchen = 5, 1, 3
ail_bath, ail_bedroom, ail_kitchen = 3, 4, 20

print("Fath baths -->", fath_bath)
print("Srutika baths -->", sru_bath)
# Fine for 3 houses. Painful for 300.
```

```python
# A class with shared attributes — every house starts the same
class House:
    bath = 4
    bedroom = 5
    kitchen = 2


fath = House()
srutika = House()
fath.bath = 5
srutika.bath = 5
print("Fath baths -->", fath.bath)          # 5
print("Srutika baths -->", srutika.bath)    # 5
print("default kitchen -->", House.kitchen)  # 2
```

```python
# __init__ sets unique values when you create each object
class House:
    def __init__(self, name, baths, bedrooms, kitchens):
        self.name = name
        self.baths = baths
        self.bedrooms = bedrooms
        self.kitchens = kitchens

    def describe(self):
        print(f"{self.name}: {self.baths} baths, {self.bedrooms} beds, {self.kitchens} kitchens")


fath = House("Fath", 5, 10, 5)
srutika = House("Srutika", 5, 1, 3)
aileen = House("Aileen", 3, 4, 20)

fath.describe()      # Fath: 5 baths, 10 beds, 5 kitchens
srutika.describe()
aileen.describe()
```

## Dataclasses

For data-shaped objects, `@dataclass` writes `__init__` for you. This is what you will see in a lot of modern Python.

```python
# dataclass — same idea, less boilerplate
from dataclasses import dataclass


@dataclass
class House:
    name: str
    baths: int
    bedrooms: int
    kitchens: int

    def describe(self) -> str:
        return f"{self.name}: {self.baths} baths, {self.bedrooms} beds, {self.kitchens} kitchens"


fath = House("Fath", 5, 10, 5)
print(fath)              # House(name='Fath', baths=5, bedrooms=10, kitchens=5)
print(fath.describe())   # Fath: 5 baths, 10 beds, 5 kitchens
```
