## Lists

```python
# Create, index, slice
courses = ["Python", "Java", "C++"]
print(courses)         # ['Python', 'Java', 'C++']
print(courses[0])      # Python
print(courses[-1])     # C++
print(courses[0:2])    # ['Python', 'Java']
print(len(courses))    # 3
```

```python
# append vs extend
courses = ["Python", "Java", "C++"]
web = ["HTML", "CSS"]

courses.extend(web)
print("after extend -->", courses)
# ['Python', 'Java', 'C++', 'HTML', 'CSS']

courses.append("SQL")
print("after append -->", courses)
# ['Python', 'Java', 'C++', 'HTML', 'CSS', 'SQL']

# append adds ONE item (even if that item is a list).
# extend adds EACH item from another list.
```

```python
# remove, pop, reverse, sort
courses = ["Python", "Java", "C++", "HTML", "CSS"]
courses.remove("Python")
print("remove -->", courses)

last = courses.pop()
print("popped -->", last)      # CSS
print("after pop -->", courses)

courses.reverse()
print("reversed -->", courses)

courses.sort()
print("sort in place -->", courses)

print("sorted() copy -->", sorted(courses, reverse=True))
print("original still -->", courses)
```

```python
# join and split
courses = ["Java", "HTML", "CSS"]
text = " - ".join(courses)
print(text)              # Java - HTML - CSS
print(text.split(" - ")) # ['Java', 'HTML', 'CSS']
```

## Tuples

```python
# A tuple uses (). You can read an item. You cannot replace one.
courses = ("History", "Math", "Physics")
print(courses[0])  # History

try:
    courses[0] = "Art"
except TypeError as err:
    print(err)  # 'tuple' object does not support item assignment

# A list uses [] and allows that same edit.
course_list = ["History", "Math", "Physics"]
course_list[0] = "Art"
print(course_list)  # ['Art', 'Math', 'Physics']

# Unpacking gives each tuple item its own name.
name, age = ("Alina", 30)
print(name, age)  # Alina 30
```

## Sets

```python
# Sets drop duplicates and have no order
courses = {"History", "Math", "Physics", "Math"}
print(courses)  # duplicates gone

courses.add("Art")
print("Art" in courses)  # True

empty_dict = {}
empty_set = set()
print(type(empty_dict))  # <class 'dict'>
print(type(empty_set))   # <class 'set'>
```

```python
# Set operations
cs = {"History", "Math", "Physics", "CompSci"}
art = {"History", "Math", "Art", "Design"}

print(cs.intersection(art))  # {'History', 'Math'}
print(cs.difference(art))    # {'Physics', 'CompSci'}
print(cs.union(art))         # all unique values from both
```

## Dictionaries

```python
# A dict maps keys to values
student = {"name": "John", "age": 25, "courses": ["Math", "CompSci"]}
print(student["name"])                    # John
print(student.get("phone"))               # None
print(student.get("phone", "Not Found"))  # Not Found

student["phone"] = "555-5555"
student.update({"name": "Jane", "age": 26})
print(student)

age = student.pop("age")
print("popped age -->", age)  # 26
```

```python
# Loop a dict
student = {"name": "Jane", "courses": ["Math", "CompSci"]}
print(student.keys())
print(student.values())
print(student.items())

for key, value in student.items():
    print(key, "-->", value)
```

```python
# Nested dicts — this is what APIs look like
payload = {
    "message": "success",
    "number": 3,
    "people": [
        {"craft": "ISS", "name": "Chris Cassidy"},
        {"craft": "ISS", "name": "Doug Hurley"},
        {"craft": "ISS", "name": "Bob Behnken"},
    ],
}

print(payload["number"])                    # 3
print(payload["people"][0]["name"])         # Chris Cassidy

for person in payload["people"]:
    print(person["name"])
```
