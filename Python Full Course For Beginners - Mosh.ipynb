```python
# Functions with Parameters
print("\n\n\nFunctions with Parameters")


def greet(name):
    print(f"Hello, {name}!")


def increment(number, by=1):
    return number + by


print("increment(1)  --> ", increment(1))  # 2
print("increment(1, 2)  --> ", increment(1, 2))  # 3
```

```python
# Functions with Variable Length Arguments
print("\n\n\nFunctions with Variable Length Arguments")


def multiply(*numbers):
    total = 1
    for number in numbers:
        total *= number
    return total


print("multiply(1, 2, 3, 4, 5)  --> ", multiply(1, 2, 3, 4, 5))  # 120
```

```python
# Functions with Keyword Arguments
print("\n\n\nFunctions with Keyword Arguments")


def increment(number, by=1):
    return number + by


print("increment(1)  --> ", increment(1))  # 2
print("increment(1, 2)  --> ", increment(1, 2))  # 3s
```
