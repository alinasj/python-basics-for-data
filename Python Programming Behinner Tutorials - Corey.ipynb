```python
courses = ["Python", "Java", "C++"]
courses_2 = ["HTML", "CSS", "JavaScript"]

courses.extend(courses_2)
print("Extend Function:", courses)

courses.append(courses_2)
print("Append Function:", courses)

courses.remove("Python")
print("Remove Function:", courses)

courses.pop()
print("Pop Function:", courses)

courses.reverse()
print("Reverse Function:", courses)

courses.sort()
print("Sort Function:", courses)

courses.sort(reverse=True)
print("Sort Function with reverse:", courses)

courses.sort(key=len)
print("Sort Function with key=len:", courses)

courses.sort(key=len, reverse=True)
print("Sort Function with key=len and reverse:", courses)

sorted_courses = sorted(courses)
print("Sorted Function:", sorted_courses)  # does not change the original list
print("Sorted Function Original list:", courses)
```

```python
# index
print(courses)
print("Index Function:", courses.index("Java"))
print("Index Function with start and end:", courses.index("Java", 0, 3))
```
