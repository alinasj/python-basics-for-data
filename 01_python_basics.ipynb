## Strings

```python
# String methods
course_name = "Python for Beginners"
print("len(course_name)        -->", len(course_name))         # 20
print("course_name.upper()     -->", course_name.upper())      # PYTHON FOR BEGINNERS
print("course_name.lower()     -->", course_name.lower())      # python for beginners
print("course_name.title()     -->", course_name.title())      # Python For Beginners
print("course_name.replace(...)-->", course_name.replace("Python", "Java"))  # Java for Beginners
print("course_name.find(...)   -->", course_name.find("Python"))  # 0
print("course_name.count(...)  -->", course_name.count("o"))   # 2
print("course_name             -->", course_name)              # original is unchanged
```

```python
# Slicing
course_name = "Python for Beginners"
print("course_name[0]      -->", course_name[0])       # P
print("course_name[0:6]    -->", course_name[0:6])     # Python
print("course_name[:6]     -->", course_name[:6])      # Python
print("course_name[11:]    -->", course_name[11:])     # Beginners
print("course_name[-4:]    -->", course_name[-4:])     # ners
```

```python
# Escape sequences
print('double quotes -->', "Python \"for\" Beginners")  # Python "for" Beginners
print("single quotes -->", "Python 'for' Beginners")    # Python 'for' Beginners
print("backslash     -->", "Python \\for\\ Beginners")  # Python \for\ Beginners
print("newline       -->", "Python\nfor\nBeginners")
```

```python
# Formatted strings (f-strings)
first = "John"
last = "Smith"
print(f"{first} [{last}] is a coder")           # John [Smith] is a coder
print(f"{len(first)} letters in {first}")       # 4 letters in John
```

```python
# Padding
first = "John"
print(first.ljust(10, "*"))   # John******
print(first.rjust(10, "*"))   # ******John
print(first.center(10, "*"))  # ***John***
```

## Numbers and operators

```python
# Arithmetic
print("10 + 3  -->", 10 + 3)   # 13
print("10 - 3  -->", 10 - 3)   # 7
print("10 * 3  -->", 10 * 3)   # 30
print("10 / 3  -->", 10 / 3)   # 3.333...  (always float)
print("10 // 3 -->", 10 // 3)  # 3         (integer division)
print("10 % 3  -->", 10 % 3)   # 1         (remainder)
print("10 ** 3 -->", 10 ** 3)  # 1000      (power)
```

```python
# Augmented assignment
x = 10
x += 3
print("x += 3 -->", x)  # 13
x -= 3
print("x -= 3 -->", x)  # 10
x *= 3
print("x *= 3 -->", x)  # 30
```

```python
# Comparison operators
x, y = 10, 20
print("x == y -->", x == y)  # False
print("x != y -->", x != y)  # True
print("x > y  -->", x > y)   # False
print("x < y  -->", x < y)   # True
print("x >= y -->", x >= y)  # False
print("x <= y -->", x <= y)  # True
```

```python
# Logical operators
x, y = True, False
print("x and y -->", x and y)  # False
print("x or y  -->", x or y)   # True
print("not x   -->", not x)    # False
```

```python
# Useful number helpers
import math

print("round(2.6)     -->", round(2.6))       # 3
print("abs(-2.6)      -->", abs(-2.6))        # 2.6
print("pow(2, 3)      -->", pow(2, 3))        # 8
print("math.ceil(2.6) -->", math.ceil(2.6))   # 3
print("math.floor(2.6)-->", math.floor(2.6))  # 2
print("math.sqrt(16)  -->", math.sqrt(16))    # 4.0
```

## Type conversion

`input()` always returns a **string**. Convert it before doing math.

Skip the next cell if you do not want the notebook to wait for typing.

```python
# input() returns a string — convert before doing math
age = input("Enter your age: ")
print("type(age)     -->", type(age))          # <class 'str'>
print("int(age) + 1  -->", int(age) + 1)
print("float(age)    -->", float(age))
print('bool("")      -->', bool(""))           # False
print("bool(0)       -->", bool(0))            # False
print("bool(1)       -->", bool(1))            # True
```

## Decisions

```python
# if / else
x, y = 10, 20
if x > y:
    print("x is greater than y")
else:
    print("x is less than y")  # this line runs
```

```python
# if / elif / else
x, y = 10, 20
if x > y:
    print("x is greater")
elif x < y:
    print("x is less")  # this line runs
else:
    print("x equals y")
```

```python
# Ternary (one-line if)
x, y = 10, 20
result = "x is greater" if x > y else "x is less"
print(result)  # x is less
```

```python
# Chained comparisons
age = 22
if 18 <= age < 65:
    print("adult, not retirement age")  # this line runs
```
