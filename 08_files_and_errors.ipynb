## pathlib

```python
from pathlib import Path

here = Path(".")
print(here.resolve())          # full path to this folder
print(Path("sales.json").name)  # sales.json
print(Path("sales.json").suffix)  # .json
```

## JSON

```python
# Write and read JSON — this is the same shape APIs return
import json
from pathlib import Path

rows = [
    {"city": "Philly", "product": "bike", "amount": 120},
    {"city": "NYC", "product": "lock", "amount": 25},
    {"city": "Philly", "product": "helmet", "amount": 40},
]

path = Path("sales.json")
path.write_text(json.dumps(rows, indent=2))
loaded = json.loads(path.read_text())
print(loaded[0]["city"])  # Philly
print(len(loaded))        # 3
```

## CSV

```python
# CSV is a table as plain text. pandas can read it too (notebook 07).
import csv
from pathlib import Path

path = Path("sales.csv")
with path.open("w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["city", "product", "amount"])
    writer.writeheader()
    writer.writerows(rows)

with path.open() as f:
    for row in csv.DictReader(f):
        print(row["city"], row["amount"])
# Philly 120
# NYC 25
# Philly 40
```

## Errors

```python
# try/except — catch the failure you expect, not every failure
from pathlib import Path

missing = Path("does_not_exist.csv")
try:
    missing.read_text()
except FileNotFoundError as err:
    print("caught -->", type(err).__name__)  # FileNotFoundError

# Raise a clear error when *your* data is wrong
amount = -5
try:
    if amount < 0:
        raise ValueError("amount cannot be negative")
except ValueError as err:
    print("caught -->", err)  # amount cannot be negative
```
