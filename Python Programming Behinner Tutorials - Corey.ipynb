```python
courses = ["Python", "Java", "C++"]
courses_2 = ["HTML", "CSS", "JavaScript"]

courses.extend(courses_2)
print("Extend Function:", courses)

courses.append(courses_2)
print("Append Function:", courses)

courses.remove("Python")
print("Remove Function:", courses)

courses.pop()
print("Pop Function:", courses)

courses.reverse()
print("Reverse Function:", courses)

courses.sort()
print("Sort Function:", courses)

courses.sort(reverse=True)
print("Sort Function with reverse:", courses)

courses.sort(key=len)
print("Sort Function with key=len:", courses)

courses.sort(key=len, reverse=True)
print("Sort Function with key=len and reverse:", courses)

sorted_courses = sorted(courses)
print("Sorted Function:", sorted_courses)  # does not change the original list
print("Sorted Function Original list:", courses)
```

```python
# index
print(courses)
print("Index Function:", courses.index("Java"))
print("Index Function with start and end:", courses.index("Java", 0, 3))
```

```python
courses_str = " - ".join(courses)
print("Join Function:", courses_str)
new_list = courses_str.split(" - ")
print("Split Function:", new_list)
```

```python
# Tupils can't be changed, they are immutable
# Tupils are faster than lists
# Tupils are used to store related data

list_1 = ["History", "Math", "Physics", "CompSci"]
list_2 = list_1
print(list_1)
list_1[0] = "Art"
print("\n\n\nPost List Assignment:")
print(list_1)
print(list_2)

# immutable
tuple_1 = ("History", "Math", "Physics", "CompSci")
tuple_2 = tuple_1

print(tuple_1)
print(tuple_2)
```

```python
# sets - they throw out duplicates
courses = {"History", "Math", "Physics", "CompSci"}
print(courses)

courses.add("Art")
print(courses)
```

```python
cs_courses = {"History", "Math", "Physics", "CompSci"}
art_courses = {"History", "Math", "Art", "Design"}

print(cs_courses.intersection(art_courses))
print(cs_courses.difference(art_courses))
print(cs_courses.union(art_courses))


empty_set = {}  # <- see this is a dictionary and not a set
empty_set_2 = set()

print(type(empty_set))
print(type(empty_set_2))
```

```python
# Dictionaries
student = {"name": "John", "age": 25, "courses": ["Math", "CompSci"]}
print(student)

student["phone"] = "555-5555"
print(student)
print(student.get("phone"))
print(student.get("phone1", "Not Found"))

student.update({"name": "Jane", "age": 26, "phone": "555-5555"})
print(student)
```

```python
age = student.pop("age")
print(age)
print(student)

student.popitem()
print(student)
```
